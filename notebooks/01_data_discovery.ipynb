## Step 4 : Data Quality Checks

```python
# 4.1 Null check — are there any unexpected blanks?
print("=== NULL COUNTS ===\n")
print("--- 2021 ---")
print(df_2021.isnull().sum())
print("\n--- 2026 ---")
print(df_2026.isnull().sum())
print("\n--- Master ---")
print(df_master.isnull().sum())
```

```python
# 4.2 Duplicate check — any row appearing twice?

# A real duplicate = same candidate, same AC, in the same year
print(f"2021 duplicate (ac_number, candidate): {df_2021.duplicated(subset=['ac_number','candidate']).sum()}")
print(f"2026 duplicate (ac_number, candidate): {df_2026.duplicated(subset=['ac_number','candidate']).sum()}")
print(f"Master duplicate (ac_number):          {df_master.duplicated(subset=['ac_number']).sum()}")


# Show the actual duplicate rows in 2026 to inspect them
print("=== 2026 duplicates ===")
dupes_26 = df_2026[df_2026.duplicated(subset=['ac_number','candidate'], keep=False)].sort_values(['ac_number','candidate'])
print(dupes_26[['ac_number','constituency','candidate','party','votes']].to_string(index=False))

print("\n\n=== 2021 duplicates ===")
dupes_21 = df_2021[df_2021.duplicated(subset=['ac_number','candidate'], keep=False)].sort_values(['ac_number','candidate'])
print(dupes_21[['ac_number','constituency','candidate','party','votes']].to_string(index=False))
```

```python
# 4.3 AC number completeness — are all 234 ACs covered?
# Expected set: {1, 2, 3, ..., 234}
expected = set(range(1, 235))

ac_21 = set(df_2021['ac_number'].unique())
ac_26 = set(df_2026['ac_number'].unique())
ac_master = set(df_master['ac_number'].unique())

print(f"Missing ACs in 2021: {sorted(expected - ac_21) or 'None ✅'}")
print(f"Missing ACs in 2026: {sorted(expected - ac_26) or 'None ✅'}")
print(f"Missing ACs in master: {sorted(expected - ac_master) or 'None ✅'}")
```

```python
# 4.4 Reserved-seat sanity — does it match the brief?
# Use drop_duplicates so each AC counts once, not once per candidate
res_21 = df_2021.drop_duplicates('ac_number')['reserved'].value_counts().to_dict()
res_26 = df_2026.drop_duplicates('ac_number')['reserved'].value_counts().to_dict()
res_master = df_master['reserved'].value_counts().to_dict()

print(f"2021 reserved counts: {res_21}")
print(f"2026 reserved counts: {res_26}")
print(f"Master reserved counts: {res_master}")
```

```python
#4.5 Party-name consistency — the silent killer
parties_21 = set(df_2021['party'].unique())
parties_26 = set(df_2026['party'].unique())

print(f"Parties in 2021 only: {len(parties_21 - parties_26)}")
print(sorted(parties_21 - parties_26))
print()
print(f"Parties in 2026 only: {len(parties_26 - parties_21)}")
print(sorted(parties_26 - parties_21))
print()
print(f"Common to both: {len(parties_21 & parties_26)}")
```

```python
# 4.6 Constituency name spelling drift — confirm we should join on ac_number
# Compare constituency names at the same ac_number across years
merged_names = df_2021[['ac_number','constituency']].drop_duplicates().merge(
    df_2026[['ac_number','constituency']].drop_duplicates(),
    on='ac_number',
    suffixes=('_2021','_2026')
)

# Show rows where the spelling differs
mismatches = merged_names[merged_names['constituency_2021'] != merged_names['constituency_2026']]
print(f"Constituencies with name-spelling drift between 2021 and 2026: {len(mismatches)}")
print()
print(mismatches.head(20))
```

```python
# 4.7 Final cross-check — does master match the result files?
# Master + 2021 join sanity
check = df_master.merge(
    df_2021.drop_duplicates('ac_number')[['ac_number','reserved','region']],
    on='ac_number',
    suffixes=('_master','_2021')
)

# Do master.reserved and 2021.reserved always agree?
reserved_match = (check['reserved_master'] == check['reserved_2021']).all()
region_match = (check['region_master'] == check['region_2021']).all()

print(f"Master.reserved matches 2021.reserved for all 234 ACs: {reserved_match}")
print(f"Master.region matches 2021.region for all 234 ACs: {region_match}")
```

## Step 5 — Build the Winners Table (the foundation)

```python
# 5.1 Define MAJOR_PARTIES and the normalization function

# the parties we track explicitly. Everything else → 'OTHERS'.
MAJOR_PARTIES = {
    'DMK',         # Dravida Munnetra Kazhagam
    'AIADMK',      # All India Anna Dravida Munnetra Kazhagam
    'TVK',         # Tamilaga Vettri Kazhagam — new 2026 entrant
    'BJP',         # Bharatiya Janata Party
    'INC',         # Indian National Congress
    'PMK',         # Pattali Makkal Katchi
    'VCK',         # Viduthalai Chiruthaigal Katchi
    'NTK',         # Naam Tamilar Katchi
    'CPI',         # Communist Party of India
    'CPI(M)',      # Communist Party of India (Marxist)
    'DMDK',        # Desiya Murpokku Dravida Kazhagam
    'AMMK',        # Amma Makkal Munnetra Kazhagam
    'MDMK',        # Marumalarchi Dravida Munnetra Kazhagam
    'IND',         # Independent candidates
    'NOTA',        # None Of The Above
}

def normalize_party(p):
    """Bucket any party not in MAJOR_PARTIES into 'OTHERS'."""
    return p if p in MAJOR_PARTIES else 'OTHERS'

# Apply to both years — create a NEW column, don't overwrite raw 'party'
df_2021['party_clean'] = df_2021['party'].apply(normalize_party)
df_2026['party_clean'] = df_2026['party'].apply(normalize_party)

print("Normalization applied. Sample of new column:")
df_2026[['constituency','candidate','party','party_clean','votes']].head(10)
```

```python
# 5.2 Verify the normalization didn't break anything

print(f"=== 2021 party_clean counts ===")
print(df_2021['party_clean'].value_counts())
print(f"\n=== 2026 party_clean counts ===")
print(df_2026['party_clean'].value_counts())
```

```python
# 5.3 Build the winners table - the function

def build_winners_table(df, year_label):
    """
    Build one-row-per-AC table with:
      winner + runner-up info, margin, regional metadata.
    """
    # Sort: lowest ac_number first, then highest votes first.
    # That puts the winner at the top of each AC's group.
    s = df.sort_values(['ac_number', 'votes'], ascending=[True, False]).copy()

    # rank: 1 = winner, 2 = runner-up, 3+ = also-rans
    s['rank'] = s.groupby('ac_number').cumcount() + 1

    # Pull out winner rows
    winners = s[s['rank'] == 1][['ac_number', 'constituency', 'reserved', 'region',
                                   'candidate', 'party', 'party_clean', 'votes']].copy()
    winners = winners.rename(columns={
        'candidate':   'winner_candidate',
        'party':       'winner_party_raw',
        'party_clean': 'winner_party',
        'votes':       'winner_votes'
    })

    # Pull out runner-up rows
    runners = s[s['rank'] == 2][['ac_number', 'candidate', 'party_clean', 'votes']].copy()
    runners = runners.rename(columns={
        'candidate':   'runner_up_candidate',
        'party_clean': 'runner_up_party',
        'votes':       'runner_up_votes'
    })

    # Join winners + runners on ac_number
    out = winners.merge(runners, on='ac_number', how='left')

    # Total valid votes in the AC (denominator for margin %)
    totals = df.groupby('ac_number')['votes'].sum().rename('total_valid_votes').reset_index()
    out = out.merge(totals, on='ac_number')

    # Margin metrics
    out['margin']                  = out['winner_votes'] - out['runner_up_votes']
    out['margin_pct']              = (out['margin'] / out['total_valid_votes']) * 100
    out['winner_vote_share_pct']   = (out['winner_votes'] / out['total_valid_votes']) * 100

    # Add the year tag (useful when we concat the two tables later)
    out['year'] = year_label

    # Order columns nicely
    col_order = ['year', 'ac_number', 'constituency', 'region', 'reserved',
                 'winner_candidate', 'winner_party', 'winner_party_raw', 'winner_votes',
                 'runner_up_candidate', 'runner_up_party', 'runner_up_votes',
                 'margin', 'margin_pct', 'winner_vote_share_pct', 'total_valid_votes']
    return out[col_order]

winners_2021 = build_winners_table(df_2021, 2021)
winners_2026 = build_winners_table(df_2026, 2026)

print(f"winners_2021 shape: {winners_2021.shape}")
print(f"winners_2026 shape: {winners_2026.shape}")
```

```python
# 5.6 Save the winners tables to processed/

# Save both years and a combined version for later use
PROCESSED = REPO / "data" / "processed"
PROCESSED.mkdir(exist_ok=True)

winners_2021.to_csv(PROCESSED / "winners_2021.csv", index=False)
winners_2026.to_csv(PROCESSED / "winners_2026.csv", index=False)

# Combined long-format version (both years stacked) — useful for charts
winners_long = pd.concat([winners_2021, winners_2026], ignore_index=True)
winners_long.to_csv(PROCESSED / "winners_combined.csv", index=False)

print(f"Saved to {PROCESSED}:")
for f in sorted(PROCESSED.iterdir()):
    print(f"  {f.name}  ({f.stat().st_size:,} bytes)")
```
